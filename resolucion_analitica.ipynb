$$
I = \int_0^1 x^2 e^{-x}\, dx.
$$

Aplicamos **integración por partes** dos veces.

### Primera integración por partes

$$
u = x^2, \qquad dv = e^{-x}\, dx,
$$

$$
du = 2x\, dx, \qquad v = -e^{-x}.
$$

$$
\int x^2 e^{-x} \; dx = -x^2 e^{-x} + \int 2x \; e^{-x} \; dx = -x^2 e^{-x} + 2\int x \; e^{-x} \; dx.
$$

### Segunda integración por partes

$$
u = x, \qquad dv = e^{-x} \; dx,
$$

$$
du = dx, \qquad v = -e^{-x}.
$$

$$
\int x\, e^{-x} \; dx = -x e^{-x} + \int e^{-x} \; dx = -x e^{-x} - e^{-x} + C.
$$

Sustituyendo

$$
\int x^2 e^{-x} \; dx = -x^2 e^{-x} + 2\left(-x e^{-x} - e^{-x}\right) + C = -\left(x^2 + 2x + 2\right) e^{-x} + C.
$$

Entre $0$ y $1$

$$
I = \left[-\left(x^2 + 2x + 2\right) e^{-x}\right]_0^1
= -\left(1 + 2 + 2\right) e^{-1} - \left(-(0 + 0 + 2)\, e^{0}\right)
= -5 e^{-1} + 2.
$$

$$
\boxed{\,I = 2 - \dfrac{5}{e} \approx 0{,}1606027941\,}
$$
